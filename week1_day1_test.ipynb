```python
import pandas as pd

# 가상의 상품 데이터 (2주차부터는 이 데이터를 웹에서 자동으로 수집합니다)
data = [
    {"상품명": "무선 마우스 A", "가격": 25000, "리뷰수": 120},
    {"상품명": "기계식 키보드 B", "가격": 89000, "리뷰수": 45},
    {"상품명": "노트북 거치대 C", "가격": 19000, "리뷰수": 310},
    {"상품명": "모니터 암 D", "가격": 62000, "리뷰수": 85},
    {"상품명": "게이밍 헤드셋 E", "가격": 55000, "리뷰수": 150},
]

df = pd.DataFrame(data)

# 자동 분류: 리뷰 80개 이상인 '인기 상품'만 골라내기
hot_items = df[df["리뷰수"] >= 80]

print("=== 전체 상품 목록 ===")
print(df)
print("\n=== 리뷰 100개 이상 인기 상품만 자동 추출 ===")
print(hot_items)

# 추출 결과를 엑셀에서 바로 열리는 CSV 파일로 자동 저장
hot_items.to_csv("hot_items_report.csv", index=False, encoding="utf-8-sig")
print("\n완료! 왼쪽 폴더 아이콘(📁)을 누르면 'hot_items_report.csv' 파일이 생성되어 있습니다.")
```

```python
# 1. 기존 가격에 20% 할인된 '할인가' 열 자동 추가 (빈칸 1 채우기)
df["할인가"] = (df["가격"] * 0.8).astype(int)

# 2. 가격이 50,000원 이하(&)이면서 리뷰수가 100개 이상인 가성비 상품만 추출 (빈칸 2, 3 채우기)
best_value = df[(df["가격"] <= 50000) & (df["리뷰수"] >= 100)]

print("=== 가성비 인기 상품 목록 ===")
print(best_value)

# 3. 반복문(for)으로 고객 발송용 자동 알림 문구 생성
print("\n=== 자동 알림 문구 출력 ===")
for idx, row in best_value.iterrows():
    print(f"[특가 알림] {row['상품명']} | 원가 {row['가격']}원 → 할인가 {row['할인가']}원 (리뷰 {row['리뷰수']}개)")

# 4. 결과를 엑셀용 CSV로 자동 저장
best_value.to_csv("best_value_report.csv", index=False, encoding="utf-8-sig")
```
